```python
import math
import pandas as pd
import matplotlib.pyplot as plt

df= pd.read_csv("cleaned_dataset_v2.3.csv")
df.head()
```

```python
df.info()
```

```python
col = df.columns

feature_col = col[2:17]
days_col = col[16:]
```

```python
def plot_graphs(df):
    num_rows = df.shape[0]           
    cols = 5
    rows = math.ceil(num_rows / cols)
    
    plt.figure(figsize=(20, 15))
    
    for i in range(0,num_rows):
        plt.subplot(rows, cols, i + 1)
        plt.plot(df.iloc[i].values, marker='o', linewidth=1)
        plt.title(f"Row {i}")
        plt.xlabel("Day")
        plt.ylabel("Shrinkage")
        plt.grid(True)
    
    plt.tight_layout()
    plt.show()
```

```python
X=df[feature_col]
y=df[days_col]
plot_graphs(y)
```

```python
X.describe()
```

```python
X['Sand/Water']=X['Sand']/X['Water']
```

```python
X.info()
```

```python
X.head()
```

```python
X=X.drop(columns=['Sand','Water','print_type_Cast','Binder_Weight'],axis=1)
```

```python
y.describe()
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(
    max_depth=4,
    random_state=42
)

model.fit(X, y)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

y_pred = model.predict(X)

print("Training MAE:", mean_absolute_error(y, y_pred))
print("Training R2:", r2_score(y, y_pred))
```

```python
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error
import numpy as np

gkf = GroupKFold(n_splits=5)

maes = []
base = []
groups = np.arange(X.shape[0])  # one group per row

for train_idx, test_idx in gkf.split(X, y, groups):
    X_tr = X.iloc[train_idx]
    X_te = X.iloc[test_idx]

    y_tr = y.iloc[train_idx]
    y_te = y.iloc[test_idx]

    baseline_pred = np.zeros_like(y_te.values)
    baseline_pred[:, 0] = y_te.values[:, 0]
    baseline_pred[:, 1:] = y_te.values[:, :-1]

    baseline_mae = mean_absolute_error(
        y_te.values.flatten(),
        baseline_pred.flatten()
    )
    base.append(baseline_mae)


    mono = [0] * (X.shape[1] - 1) + [-1]  # example

    model = XGBRegressor(
        n_estimators=300,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        monotone_constraints=mono,
        objective="reg:absoluteerror",
        random_state=42
    )

    model.fit(X_tr, y_tr)
    y_pred = model.predict(X_te)


    model.fit(X_tr, y_tr)
    y_pred = model.predict(X_te)

    # enforce monotonicity
    y_pred = np.minimum.accumulate(
        y_pred.reshape(y_te.shape), axis=1
    )

    mae = mean_absolute_error(
    y_te.values.flatten(),
    y_pred.flatten()
)

    maes.append(mae)

print("Fold MAEs:", maes)
print("CV MAE:", np.mean(maes))
```
