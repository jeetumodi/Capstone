```python
import math
import pandas as pd
import matplotlib.pyplot as plt

df= pd.read_csv("cleaned_dataset_v2.3.csv")
df.head()
```

```python
col = df.columns

feature_col = col[2:17]
days_col = col[16:]
```

```python
df['Sand/Water']=df['Sand']/df['Water']
```

```python
df=df.drop(columns=['Sand','Water','print_type_Cast','Nano-Silica','Fly_ash','Ultrafine_Fly_ash','Ordinary_Portland_Cement'],axis=1)
```

```python
df.info()
```

```python
target_cols = [f"day_{i:02d}" for i in range(1, 31)]

X = df.drop(columns=target_cols + ["Mix_name"])
y = df[target_cols]
```

```python
from xgboost import XGBRegressor
from sklearn.multioutput import MultiOutputRegressor

base_model = XGBRegressor(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:absoluteerror",
    random_state=42
)

model = MultiOutputRegressor(base_model)
model.fit(X, y)
```

```python
from sklearn.isotonic import IsotonicRegression
import numpy as np

ir = IsotonicRegression(increasing=False)

y_pred = model.predict(X)

y_pred_mono = np.vstack([
    ir.fit_transform(np.arange(30), row)
    for row in y_pred
])
```

```python
import joblib

joblib.dump(
    {
        "model": model,         
        "isotonic": ir           
    },
    "xgb_multioutput_isotonic.pkl"
)
```

```python
from sklearn.metrics import mean_absolute_error

mae_raw = mean_absolute_error(y.values.flatten(), y_pred.flatten())
mae_mono = mean_absolute_error(y.values.flatten(), y_pred_mono.flatten())

print("Raw MAE:", mae_raw)
print("Monotonic MAE:", mae_mono)
```

```python
y_pred=pd.DataFrame(y_pred_mono)
```

```python
num_rows = y.shape[0]           
cols = 5
rows = math.ceil(num_rows / cols)
    
plt.figure(figsize=(20, 15))
    
for i in range(0,num_rows):
    plt.subplot(rows, cols, i + 1)
   # Plot the actual data in blue with circles
    plt.plot(y.iloc[i].values, marker='o', linewidth=1, color='blue', label='Actual')
    plt.plot(y_pred.iloc[i].values, marker='+', linewidth=1, color='red', label='Predicted')
    plt.title(f"Row {i}")
    plt.xlabel("Day")
    plt.ylabel("Shrinkage")
    plt.grid(True)
    
plt.tight_layout()
plt.show()
```

```python
import numpy as np

y_delta = y.values[:, :-1] - y.values[:, 1:]
y_delta = np.maximum(y_delta, 0)  # safety
```

```python
from xgboost import XGBRegressor
from sklearn.multioutput import MultiOutputRegressor

delta_model = MultiOutputRegressor(
    XGBRegressor(
        n_estimators=50,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:absoluteerror",
        random_state=42
    )
)

delta_model.fit(X, y_delta)
```

```python

joblib.dump(delta_model, "delta_xgb_multioutput.pkl")
```

```python
delta_pred = np.maximum(0, delta_model.predict(X))

y0 = y.values[:, [0]]  # day_01

y_rest = y0 - np.cumsum(delta_pred, axis=1)

y_pred = np.hstack([y0, y_rest])
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(
    y.values.flatten(),
    y_pred.flatten()
)

print("MAE:", mae)
```

```python
num_rows = y.shape[0]           
cols = 5
rows = math.ceil(num_rows / cols)
    
plt.figure(figsize=(20, 15))
    
for i in range(0,num_rows):
    plt.subplot(rows, cols, i + 1)
    plt.plot(y.iloc[i].values, marker='o', linewidth=1, color='blue', label='Actual')
    plt.plot(y_pred[i], marker='+', linewidth=1, color='red', label='Predicted')
    plt.title(f"Row {i}")
    plt.xlabel("Day")
    plt.ylabel("Shrinkage")
    plt.grid(True)
    
plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import ElasticNet

model = ElasticNet(alpha=0.01, l1_ratio=0.7)
model.fit(X, y_delta)
delta_pred = np.maximum(model.predict(X), 0)
```

```python
y0 = y.values[:, [0]]  # day_01
y_rest = y0 - np.cumsum(delta_pred, axis=1)
y_pred = np.hstack([y0, y_rest])
```

```python

joblib.dump(delta_model, "delta_esaltic.pkl")
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(
    y.values.flatten(),
    y_pred.flatten()
)

print("MAE:", mae)
```
