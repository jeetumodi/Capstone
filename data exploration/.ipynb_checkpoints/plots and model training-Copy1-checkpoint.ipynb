```python
import math
import pandas as pd
import matplotlib.pyplot as plt

df= pd.read_csv("cleaned_dataset_v2.3.csv")
df.head()
```

```python
df.info()
```

```python
col = df.columns

feature_col = col[2:17]
days_col = col[16:]
```

```python
def plot_graphs(df):
    num_rows = df.shape[0]           
    cols = 5
    rows = math.ceil(num_rows / cols)
    
    plt.figure(figsize=(20, 15))
    
    for i in range(0,num_rows):
        plt.subplot(rows, cols, i + 1)
        plt.plot(df.iloc[i].values, marker='o', linewidth=1)
        plt.title(f"Row {i}")
        plt.xlabel("Day")
        plt.ylabel("Shrinkage")
        plt.grid(True)
    
    plt.tight_layout()
    plt.show()
```

```python
X=df[feature_col]
y=df[days_col]
plot_graphs(y)
```

```python
X.describe()
```

```python
X.info()
```

```python
X.head()
```

```python
X=X.drop(columns=['Sand','Water'],axis=1)
```

```python
y.describe()
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(
    max_depth=4,
    random_state=42
)

model.fit(X, y)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

y_pred = model.predict(X)

print("Training MAE:", mean_absolute_error(y, y_pred))
print("Training R2:", r2_score(y, y_pred))
```

```python
type(y_pred)
```

```python
y_pred.shape
```

```python
y_pre=pd.DataFrame(y_pred)
```

```python
num_rows = y.shape[0]           
cols = 5
rows = math.ceil(num_rows / cols)
    
plt.figure(figsize=(20, 15))
    
for i in range(0,num_rows):
    plt.subplot(rows, cols, i + 1)
   # Plot the actual data in blue with circles
    plt.plot(y.iloc[i].values, marker='o', linewidth=1, color='blue', label='Actual')
    plt.plot(y_pre.iloc[i].values, marker='+', linewidth=1, color='red', label='Predicted')
    plt.title(f"Row {i}")
    plt.xlabel("Day")
    plt.ylabel("Shrinkage")
    plt.grid(True)
    
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

residuals = y_pred - y  # shape: (rows, days)
mean_residual = residuals.mean(axis=0)

plt.figure(figsize=(20,4))
plt.plot(mean_residual, marker='o')
plt.axhline(0)
plt.xlabel("Day")
plt.ylabel("Mean Residual")
plt.title("Residuals vs Day")
plt.grid(True)
plt.show()
```

```python
from xgboost import XGBRegressor

model = XGBRegressor(
    n_estimators=400,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

model.fit(X, y)


y_pred = model.predict(X)

print("Training MAE:", mean_absolute_error(y, y_pred))
print("Training R2:", r2_score(y, y_pred))
y_pre=pd.DataFrame(y_pred)

num_rows = y.shape[0]           
cols = 5
rows = math.ceil(num_rows / cols)
    
plt.figure(figsize=(20, 15))
    
for i in range(0,num_rows):
    plt.subplot(rows, cols, i + 1)
   # Plot the actual data in blue with circles
    plt.plot(y.iloc[i].values, marker='o', linewidth=1, color='blue', label='Actual')
    plt.plot(y_pre.iloc[i].values, marker='+', linewidth=1, color='red', label='Predicted')
    plt.title(f"Row {i}")
    plt.xlabel("Day")
    plt.ylabel("Shrinkage")
    plt.grid(True)
    
plt.tight_layout()
plt.show()
```
